## Dot Product Example (Nut Butter Store Sales)

```python
np.random.seed(1)
sales_amounts = np.random.randint(10, size = (5, 3))
sales_amounts
```

```python
# Create weekly sales DataFrame
import pandas as pd
weekly_sales = pd.DataFrame(sales_amounts, index=("Mon", "Tues", "Wed", "Thurs", "Fri"),
                            columns=("Almond butter", "Peanut butter", "Cashew butter"))
weekly_sales
```

```python
prices = np.array([10, 8, 12])
prices
```

```python
# Create butter_prices dataFrame
butter_prices = pd.DataFrame(prices.reshape(1, 3),
                      index=["Price"], columns=["Almond butter", "Peanut butter", "Cashew butter"])
```

```python
butter_prices.shape
```

```python
total_sales = np.dot(prices, sales_amounts.T)# prices.dot(sales_amounts.T)
total_sales
```

```python
weekly_sales
```

```python
Total_sales = np.array(total_sales)
Total_sales
```

```python
sales_amounts.shape
```

```python
weekly_sales["Total_Sales"] = Total_sales
weekly_sales
```

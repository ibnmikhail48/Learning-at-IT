## Matplotlib example workflow

```python
# There are six steps to using matplotlib

# 0. Import matplotlib and get it ready for plotting in Jupyter
%matplotlib inline
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# 1. Prepare data
x = [12, 15, 18, 21, 24]
y = [21, 28, 35, 42, 49]


# 2. Setup plot
fig, ax = plt.subplots()

# 3. Plot the data
ax.plot(x, y);


# 4. Customize the plot
ax.set(title="The Plot of X axis against Y axis", xlabel="x axis", ylabel="y axis")


# 5. Set the Legend
ax.legend([x, y])


plt.show()
```

```python
# 0. Import matplotlib and get it ready for plotting in Jupyter
%matplotlib inline
import matplotlib.pyplot as plt

# 1.Prepare data
x = [1, 2, 3, 4]
y = [11, 22, 33, 44]

# 2. Setup plot
fig, ax = plt.subplots(figsize=(5, 5)) # Heigh and weight

# 3. Plot data
ax.plot(x,y, marker="o")

# 4. Customize plot
ax.set(title="Simple Plot", xlabel="x-axis", ylabel="y-axis", )

# 5. Save and show (you save the whole figure)
fig.savefig("sample-plot.png")


plt.show()


```

## Two options for subplots

```python
# Subplots option 1 Tuple
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows=2, ncols=2, figsize=(10,5));
ax1.plot(x, x/2)
ax2.scatter(np.random.random(10), np.random.random(10))
ax3.bar(nut_butter_prices.keys(), nut_butter_prices.values());
ax4.hist(np.random.randn(1000));

plt.show()
```

```python
# Subplots option 2
fig, ax = plt.subplots(nrows=2,
                       ncols=2,
                       figsize = (10, 5))
ax[0, 0].plot(x, x/2)
ax[0, 1].scatter(np.random.random(10), np.random.random(10))
ax[1, 0].bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax[1, 1].hist(np.random.randn(1000))

plt.show()
```

## Which one should you use? (pyplot vs matplotlib OO method)
* When plotting something quickly, okay to use the pyplot method
* When plotting something more advanced, use matplotlib OO method

```python
heart_disease.head()
```

```python
over_50 = heart_disease[heart_disease["age"] > 50]
over_50.head()
```

```python
# Pyplot Method
over_50.plot(kind="scatter", x="age", y="chol", c="target");
plt.show();
```

```python
# OO method mixed with pyplot method
fix, ax = plt.subplots(figsize=(10, 6))
over_50.plot(kind="scatter", x="age", y="chol", c="target", ax=ax);


# ax.set_xlim([45, 100]);
plt.show();
```

```python
over_50.target.values
```

```python
# OO method from scratch
fig, ax = plt.subplots(figsize=(10, 5))
scatter = ax.scatter(x=over_50["age"], y=over_50["chol"], c=over_50["target"]);

# customize the plot
ax.set(title="Heart Disease and Cholesterol Levels", xlabel="Age", ylabel="Cholesterol");

#Add a legend
ax.legend(*scatter.legend_elements(), title="Target");

# Add a horizontal line
ax.axhline(over_50["chol"].mean(), linestyle="-.");
plt.show();
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
# Create a subplot of chol, age and thalach
fig, (ax0, ax1) = plt.subplots(nrows=2, ncols=1, figsize=(10, 10),
                              sharex = True)

# Add data to ax0
scatter = ax0.scatter( x=over_50["age"],
                      y=over_50["chol"],
                      c=over_50["target"])
#Customize ax0
ax0.set(title="Heart Disease and Cholesterol Levels",
        ylabel= "Cholesterol")
# Add legend

ax0.legend(*scatter.legend_elements(), title="Target")

# Add horizontal line

ax0.axhline(y=over_50["chol"].mean(), linestyle="-.")


# Add data to ax1
scatter = ax1.scatter( x=over_50["age"],
                      y=over_50["thalach"],
                      c=over_50["target"]);

# Customize ax1
ax1.set(title="Heart Disease and Max Heart Rate",
        xlabel="Age",
        ylabel="Max Heart Rate");

# Add legend
ax1.legend(*scatter.legend_elements(), title="Target")

# Add a horizontal line
ax1.axhline(y=over_50["thalach"].mean(), linestyle=":")

# Add a title to the whole figure

fig.suptitle("Heart Disease Analysis", fontsize=16, fontweight="bold");

# Save the plot

plt.savefig("Heart Disease Analysis");




plt.show();
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
# Customizing the x and y axis limitattion
# Create a subplot of chol, age and thalach
fig, (ax0, ax1) = plt.subplots(nrows=2, ncols=1, figsize=(10, 10),
                              sharex = True)

# Add data to ax0
scatter = ax0.scatter( x=over_50["age"],
                      y=over_50["chol"],
                      c=over_50["target"],
                     cmap="winter")
#Customize ax0
ax0.set(title="Heart Disease and Cholesterol Levels",
        ylabel= "Cholesterol")
# change the x axis limit
ax0.set_xlim([50, 80]);
# Add legend

ax0.legend(*scatter.legend_elements(), title="Target")

# Add horizontal line

ax0.axhline(y=over_50["chol"].mean(), c="r", linestyle="-.")


# Add data to ax1
scatter = ax1.scatter( x=over_50["age"],
                      y=over_50["thalach"],
                      c=over_50["target"],
                     cmap="winter");

# Customize ax1
ax1.set(title="Heart Disease and Max Heart Rate",
        xlabel="Age",
        ylabel="Max Heart Rate");
# Change ax1 x axis limits
ax1.set_xlim([50, 80]);
# Change ax1 y axis limits
ax1.set_ylim([60, 200]);

# Add legend
ax1.legend(*scatter.legend_elements(), title="Target")

# Add a horizontal line
ax1.axhline(y=over_50["thalach"].mean(), c="r", linestyle=":")

# Add a title to the whole figure

fig.suptitle("Heart Disease Analysis", fontsize=16, fontweight="bold");

# Save the plot

plt.savefig("Heart Disease Analysis");

plt.savefig("Heart Disease Analysis Plot")



plt.show();
```
